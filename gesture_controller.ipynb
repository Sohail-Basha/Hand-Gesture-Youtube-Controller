```python
import mediapipe as mp
import numpy
import cv2
import pyautogui as pag
import time
import math
```

```python
handsign=mp.solutions.hands
gesture=handsign.Hands(static_image_mode=False,max_num_hands=1,min_detection_confidence=0.6,min_tracking_confidence=0.6)
handsmark=mp.solutions.drawing_utils
```

```python
finger_tip=[4,8,12,16,20]
last_action_time=0
action_delay=2
```

```python
video=cv2.VideoCapture(0)
while video.isOpened():
    success,frame=video.read()
    if not success:
        break
    rgb_image=cv2.cvtColor(frame,cv2.COLOR_BGR2RGB)
    result=gesture.process(rgb_image)
    if result.multi_hand_landmarks:
        for hand in result.multi_hand_landmarks:
            handsmark.draw_landmarks(frame,hand,handsign.HAND_CONNECTIONS)
            listt=[]
            for id,lm in enumerate(hand.landmark):
                h,w,c=frame.shape
                cx,cy=int(lm.x*w),int(lm.y*h)
                listt.append([id,cx,cy])
            x_list = [pt[1] for pt in listt]
            y_list = [pt[2] for pt in listt]
            xmin,xmax=min(x_list),max(x_list)
            ymin,ymax=min(y_list),max(y_list)
            cv2.rectangle(frame,pt1=(xmin,ymin),pt2=(xmax,ymax),color=(0,255,0),thickness=2)
            
            fingers=[]
            if listt[4][1] > listt[3][1]:
                fingers.append(1)
            else:
                fingers.append(0)
            for tip in finger_tip[1:]:
                if listt[tip][2] < listt[tip - 2][2]:
                    fingers.append(1)
                else:
                    fingers.append(0)
            totalFingers = fingers.count(1)
        
            cv2.putText(frame,f"fingers:{totalFingers}",(xmin,ymin),cv2.FONT_HERSHEY_SIMPLEX,0.9,(0,255,0),2)
            current_time=time.time()
            if current_time - last_action_time>action_delay:
                if totalFingers == 5:
                    pag.press('space')   # Pause
                    last_action_time = current_time

                elif totalFingers == 2:
                    pag.press('right')   # Forward
                    last_action_time = current_time

                elif totalFingers == 3:
                    pag.press('left')    # Backward
                    last_action_time = current_time
                elif totalFingers == 4:
                    pag.press('volumeup')
                    last_action_time = current_time #volumeup
                elif totalFingers == 0:
                    pag.press('volumedown')
                    last_action_time = current_time #volumedown
                elif totalFingers == 1:
                    pag.press('f') #fullscreen
    cv2.imshow("WebCam",frame)
    if cv2.waitKey(1) & 255 == ord('q'):
        break
video.release()
cv2.destroyAllWindows()
```
